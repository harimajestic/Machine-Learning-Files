```python
import pandas as pd
from sklearn.model_selection  import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute  import SimpleImputer
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
df= pd.read_csv('application_train.csv')
```

```python
x=df.drop('TARGET',axis=1)
y=df['TARGET']
```

```python
num_cal=x.select_dtypes(exclude='object').columns
cat_cal=x.select_dtypes(include='object').columns
```

```python

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
num_pipeline=Pipeline([('SimpleImputer',SimpleImputer(strategy='median'))])
```

```python
cat_pipeline=Pipeline([('SimpleImputer',SimpleImputer(strategy='most_frequent')),('onehotencoder',OneHotEncoder(handle_unknown='ignore',sparse_output=True))])
```

```python
preprocessor=ColumnTransformer([('num_pipeline',num_pipeline,num_cal),('cat_pipeline',cat_pipeline,cat_cal)])
```

```python
model_pipeline=Pipeline([('preprocessor',preprocessor),('randomforestclassifier',RandomForestClassifier(n_estimators=6))])
```

```python
model_pipeline.fit(x_train,y_train)
```

```python
train_predict=model_pipeline.predict(x_train)
```

```python
accuracy_score(y_train,train_predict)
```

```python
test_predict=model_pipeline.predict(x_test)
```

```python
accuracy_score(test_predict,y_test)
```
